```sql
%python

import os
import shutil

src_volume_path = "/Volumes/dbacademy_retail/v01/retail-pipeline"
tgt_volume_path = "/Volumes/db_retail/dbacademy_retail_v01/retail-pipeline"


def copy_files_from_volume_to_volume(src_path, tgt_path, n):
    if n <= 0:
        print(f"Skipping copy for {src_path}: n={n}")
        return []

    if not os.path.exists(src_path):
        raise FileNotFoundError(f"Source directory does not exist: {src_path}")

    # Filter to regular files and sort so n=1 consistently picks the first file
    files = sorted(
        f for f in os.listdir(src_path)
        if os.path.isfile(os.path.join(src_path, f))
    )

    if not files:
        print(f"No files found in {src_path}")
        return []

    files_to_copy = files[:n]
    os.makedirs(tgt_path, exist_ok=True)

    copied_files = []
    for file in files_to_copy:
        src_file = os.path.join(src_path, file)
        tgt_file = os.path.join(tgt_path, file)
        shutil.copy2(src_file, tgt_file)
        copied_files.append(tgt_file)

    for file in copied_files:
        src_file = os.path.join(src_path, os.path.basename(file))
        print(f"Copied '{src_file}' to '{file}'")
    return copied_files


# Orders (e.g., n=1 copies the first file, n=2 copies the first two, etc.)
copy_files_from_volume_to_volume(
    src_path=f"{src_volume_path}/orders/stream_json",
    tgt_path=f"{tgt_volume_path}/orders/stream_json",
    n=3,
)

# Status
copy_files_from_volume_to_volume(
    src_path=f"{src_volume_path}/status/stream_json",
    tgt_path=f"{tgt_volume_path}/status/stream_json",
    n=3,
)

# Customers
copy_files_from_volume_to_volume(
    src_path=f"{src_volume_path}/customers/stream_json",
    tgt_path=f"{tgt_volume_path}/customers/stream_json",
    n=3,
)
```
