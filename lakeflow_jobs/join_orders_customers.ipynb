```sql
%sql
CREATE OR REPLACE TABLE db_retail.silver.customers_orders_silver AS
SELECT
  c.customer_id,
  c.customer_name,
  c.state,
  c.city,
  o.order_number,
  o.order_datetime,
  o.number_of_line_items,
  o.promo_info
FROM
  db_retail.bronze.customers_bronze c
JOIN
  db_retail.bronze.orders_bronze o
ON
  c.customer_id = o.customer_id;
```
